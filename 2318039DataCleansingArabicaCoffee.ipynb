# **D. Standarisasi Tipe Data Numerik (Bag Weight & Altitude)**

## **1. Ekstraksi angka berat bag dari teks '35 kg' -> 35.0**

```python
def convert_bag_weight(val):
  if pd.isna(val) or str(val).lower() == 'unknown':
    return np.nan
  val_str = str(val).lower().strip()

  numbers = re.findall(r'\d+(?:\.\d+)?', val_str)
  if not numbers:
    return np.nan

  weight = float(numbers[0])

  #Konversi ke KG
  if 'lbs' in val_str or 'pound' in val_str:
    weight = weight * 0.45359237
  elif 'g' in val_str and 'kg' not in val_str:
    weight = weight / 1000.0

  return weight


if 'Bag Weight' in df.columns:
  df['Bag Weight'] = df['Bag Weight'].apply(convert_bag_weight)

print('Standarisasi Bag Weight selesai!')
display(df[['Bag Weight']].head())
```

Standarisasi data pada kolom Bag Weight yang memiliki teks seperti "35 kg" menjadi angka utamanya saja dengan menjadikannya nilai float 35.0.

## **2. Konversi Altitude yang berupa range ('1700-1930') menjadi rerata numerik**

```python
def extract_altitude_mean(val):
    if pd.isna(val):
        return np.nan

    numbers = [float(s) for s in re.findall(r'\d+', str(val))]
    if len(numbers) >= 2:
        return sum(numbers[:2]) / 2
    elif len(numbers) == 1:
        return numbers[0]
    return np.nan

if 'Altitude' in df.columns:
    df['Altitude'] = df['Altitude'].apply(extract_altitude_mean)

print("Standarisasi Altitude selesai!")
display(df[['Altitude']].head())
```

Standarisasi data pada kolom Altitude yang memiliki data rentan seperti 1700-1930, kemudian mengambil data tengah-tengah atau mean, dari yang 1700-1930 nilai tengahnya 1815.

# **E. Standarisasi Tipe Data Tanggal (Grading Date & Expiration)**

## **1. Membuat Fungsi untuk Standarisasi, Pembersihan, dan Konversi ke pd.to_datetime**

```python
date_cols = ['Grading Date', 'Expiration']

for col in date_cols:
  if col in df.columns:
    df[col] = pd.to_datetime(df[col], errors='coerce', format='mixed')

print('Format tanggal berhasil diubah ke datetime64!')
display(df[date_cols].dtypes)
display(df[date_cols].head())
```

Standarisasi kolom Granding Date dan Expiration yang datanya merupakan tanggal dari st, nd, rd, th. Dijadikan tanggal dengan format YYYY-MM-DD

# **F. Pembersihan Aksara Non-Latin**

Standarisasi kolom yang datanya memiliki aksara non-Latin, dengan menyisakan datanya berupa alfabet Latin

## **1. Membuat Fungsi untuk Menghapus Aksara Non-Latin dan Mempertahankan Karakter Latin**

```python
def remove_non_latin(text):
  if pd.isna(text) or not isinstance(text, str):
    return text

  # Menghapus aksara non-Latin
  cleaned_text = re.sub(r'[^\x00-\x7F]+', '', text)
  cleaned_text = ' '.join(cleaned_text.split())

  # Jika string jadi kosong setelah dihapus, ganti ke Unknown
  return cleaned_text if cleaned_text else 'Unknown'


# Menerapkan pada semua kolom teks (object)
text_cols = df.select_dtypes(include=['object']).columns
for col in text_cols:
  df[col] = df[col].apply(remove_non_latin)

print('Pembersihan karakter non-Latin selesai!')

display(df[text_cols].head())
```

# **G. Pembersihan Spasi & Deduplikasi Data**

## **1. Membersihkan spasi berlebih di awal dan akhir teks pada semua kolom string**

```python
object_cols = df.select_dtypes(include=['object']).columns
for col in object_cols:
    df[col] = df[col].astype(str).str.strip()
```
